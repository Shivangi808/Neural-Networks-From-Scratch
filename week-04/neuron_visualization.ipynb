# Artificial Neuron Visualization

An artificial neuron receives inputs, multiplies them by weights,
adds a bias, and produces an output.

$$
z = w^Tx + b
$$

```python
import matplotlib.pyplot as plt
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 6))

# Input positions
inputs = [
    ("x₁", 0.5, 4.5),
    ("x₂", 0.5, 3),
    ("x₃", 0.5, 1.5)
]

# Neuron position
neuron_x = 5
neuron_y = 3

# Output position
output_x = 9
output_y = 3

# Draw input nodes
for label, x, y in inputs:
    ax.scatter(x, y, s=1200)
    ax.text(x, y, label, ha="center", va="center", fontsize=14)

# Draw neuron
ax.scatter(neuron_x, neuron_y, s=1800)
ax.text(
    neuron_x,
    neuron_y,
    "Σ + b",
    ha="center",
    va="center",
    fontsize=14
)

# Draw output node
ax.scatter(output_x, output_y, s=1200)
ax.text(
    output_x,
    output_y,
    "z",
    ha="center",
    va="center",
    fontsize=14
)

# Connect inputs to neuron
for label, x, y in inputs:
    ax.annotate(
        "",
        xy=(neuron_x - 0.5, neuron_y),
        xytext=(x + 0.5, y),
        arrowprops=dict(arrowstyle="->", linewidth=1.5)
    )

# Add weight labels
ax.text(2.5, 4.0, "w₁", fontsize=13)
ax.text(2.5, 3.0, "w₂", fontsize=13)
ax.text(2.5, 2.0, "w₃", fontsize=13)

# Connect neuron to output
ax.annotate(
    "",
    xy=(output_x - 0.5, output_y),
    xytext=(neuron_x + 0.8, neuron_y),
    arrowprops=dict(arrowstyle="->", linewidth=1.5)
)

# Labels
ax.text(0.5, 5.3, "Inputs", ha="center", fontsize=14)
ax.text(5, 5.3, "Artificial Neuron", ha="center", fontsize=14)
ax.text(9, 4.0, "Output", ha="center", fontsize=14)

ax.set_xlim(0, 10)
ax.set_ylim(0.5, 6)
ax.axis("off")

plt.title("Artificial Neuron", fontsize=18)
plt.show()
```

## How the Neuron Works

Each input is multiplied by its corresponding weight.

The weighted inputs are combined using a dot product:

$$
w^Tx = w_1x_1 + w_2x_2 + w_3x_3
$$

The bias is then added:

$$
z = w^Tx + b
$$

The result `z` is the raw output of the neuron.
