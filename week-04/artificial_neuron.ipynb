# Artificial Neuron From Scratch

## Objective

Implement the basic computation of an artificial neuron using NumPy.

The neuron calculates:

z = wᵀx + b

```python
import numpy as np
```

```python
x = np.array([2, 3, 4])

w = np.array([0.5, 0.2, 0.8])

b = 1
```

```python
z = np.dot(x, w) + b

print("Neuron output:", z)
```

```python
manual = (2 * 0.5) + (3 * 0.2) + (4 * 0.8) + 1

print("Manual calculation:", manual)
print("NumPy calculation:", z)
```

```python
assert manual == z
```

## Observation

The NumPy dot product performs:

w₁x₁ + w₂x₂ + w₃x₃

The bias is then added to obtain:

z = wᵀx + b
