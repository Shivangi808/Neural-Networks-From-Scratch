# Derivatives

## Objective

Understand derivatives as rates of change and visualize them using NumPy and Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Function

We will use:

f(x) = x²

```python
x = np.linspace(-5, 5, 200)
y = x**2

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("f(x) = x²")
plt.grid()
plt.show()
```

## Numerical Derivative

NumPy's `gradient()` can approximate the derivative numerically.

```python
dy = np.gradient(y, x)

plt.plot(x, dy)
plt.xlabel("x")
plt.ylabel("f'(x)")
plt.title("Derivative of x²")
plt.grid()
plt.show()
```

## Observation

For:

f(x) = x²

the analytical derivative is:

f'(x) = 2x

The derivative is:

- Negative when x < 0
- Zero when x = 0
- Positive when x > 0

```python
functions = {
    "x²": x**2,
    "x³": x**3,
    "sin(x)": np.sin(x)
}

for name, y_value in functions.items():
    plt.plot(x, y_value, label=name)

plt.legend()
plt.grid()
plt.title("Different Functions")
plt.show()
```
