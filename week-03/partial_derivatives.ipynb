# Partial Derivatives

## Objective

Understand how a function with multiple variables changes with respect to one variable while keeping the others constant.

We will use:

f(x,y) = x² + y²

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x, y):
    return x**2 + y**2
```

```python
#Partial derivative with respect to x
x = np.linspace(-5, 5, 100)
y = 2

df_dx = 2 * x

plt.plot(x, df_dx)
plt.xlabel("x")
plt.ylabel("∂f/∂x")
plt.title("Partial Derivative with respect to x")
plt.grid()
plt.show()
```

```python
#Partial derivative with respect to y
y = np.linspace(-5, 5, 100)

df_dy = 2 * y

plt.plot(y, df_dy)
plt.xlabel("y")
plt.ylabel("∂f/∂y")
plt.title("Partial Derivative with respect to y")
plt.grid()
plt.show()
```

## Key Takeaway

For:

f(x,y) = x² + y²

∂f/∂x = 2x

∂f/∂y = 2y

Each partial derivative measures the effect of one variable while treating the other variable as constant.
