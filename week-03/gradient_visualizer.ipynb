# Gradient Visualizer

## Objective

Visualize a function, its gradient, and understand why gradient descent moves in the opposite direction of the gradient.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-5, 5, 200)

y = x**2

gradient = 2*x
```

```python
#Plot
plt.plot(x, y, label="f(x) = x²")

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Function and Gradient")
plt.grid()
plt.legend()

plt.show()
```

```python
#Show gradient
plt.plot(x, gradient, label="Gradient = 2x")

plt.axhline(0, linewidth=0.8)

plt.xlabel("x")
plt.ylabel("Gradient")
plt.title("Gradient of x²")
plt.grid()
plt.legend()

plt.show()
```

## Observation

The gradient tells us the direction in which the function increases most rapidly.

Gradient descent moves in the opposite direction.

For a loss function:

new_parameter = old_parameter - learning_rate × gradient
