# Matrix Multiplication using NumPy

## Objective

Learn how matrix multiplication works mathematically and how it is implemented using NumPy.

---

### Topics Covered

- Creating matrices
- Matrix shapes
- Matrix multiplication
- Identity matrix
- Transpose
- Applications in Machine Learning

```python
import numpy as np
```

```python
A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

print("Matrix A")
print(A)

print("\nMatrix B")
print(B)
```

```python
#Matrix Shape
print("Shape of A:", A.shape)
print("Shape of B:", B.shape)
```

```python
#Matrix Multiplication
C = np.matmul(A, B)
print(C)
```

```python
C = A @ B

print(C)
```

### Why two methods?

`np.matmul()` and `@` perform exactly the same matrix multiplication.

The `@` operator is simply cleaner to read.

```python
#Identity Matrix
I = np.eye(3)
print(I)
```

```python
#Zero Matrix
Z = np.zeros((3,3))
print(Z)
```

```python
#Transpose
print(A.T)
```

# Matrix Multiplication in Neural Networks

A neural network computes

Z = XW + b

where

- X = Input Matrix
- W = Weight Matrix
- b = Bias

Without matrix multiplication, neural networks cannot perform forward propagation.
