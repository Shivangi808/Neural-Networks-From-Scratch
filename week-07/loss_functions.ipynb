Import NumPy

```python
import numpy as np
```

Mean Squared Error

```python
def mse(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)
```

Test MSE

```python
y_true = np.array([10, 20, 30])
y_pred = np.array([8, 18, 32])

print("MSE:", mse(y_true, y_pred))
```

Mean Absolute Error

```python
def mae(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))
```

Test MAE

```python
print("MAE:", mae(y_true, y_pred))
```

Binary Cross Entropy

```python
def binary_cross_entropy(y_true, y_pred):
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

    return -np.mean(
        y_true * np.log(y_pred) +
        (1 - y_true) * np.log(1 - y_pred)
    )
```

Test BCE

```python
y_true_binary = np.array([1, 0, 1, 1])
y_pred_binary = np.array([0.9, 0.1, 0.8, 0.7])

print(
    "Binary Cross Entropy:",
    binary_cross_entropy(y_true_binary, y_pred_binary)
)
```

Categorical Cross Entropy

```python
def categorical_cross_entropy(y_true, y_pred):
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

    return -np.mean(
        np.sum(y_true * np.log(y_pred), axis=1)
    )
```

Test Categorical Cross Entropy

```python
y_true_multi = np.array([
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1]
])

y_pred_multi = np.array([
    [0.8, 0.1, 0.1],
    [0.2, 0.7, 0.1],
    [0.1, 0.2, 0.7]
])

print(
    "Categorical Cross Entropy:",
    categorical_cross_entropy(y_true_multi, y_pred_multi)
)
```

Compare all losses

```python
print("MSE:", mse(y_true, y_pred))
print("MAE:", mae(y_true, y_pred))
print(
    "Binary Cross Entropy:",
    binary_cross_entropy(y_true_binary, y_pred_binary)
)
print(
    "Categorical Cross Entropy:",
    categorical_cross_entropy(y_true_multi, y_pred_multi)
)
```
