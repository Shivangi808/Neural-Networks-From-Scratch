Import NumPy

```python
import numpy as np
```

Define Sigmoid

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

Define Forward Pass

```python
def forward_pass(x, weights, bias):
    z = np.dot(x, weights) + bias
    prediction = sigmoid(z)

    return z, prediction
```

Create input, weights and bias

```python
x = np.array([2, 3])

weights = np.array([0.5, 0.2])

bias = 1
```

Perform forward propagation

```python
z, prediction = forward_pass(x, weights, bias)

print("Weighted Sum:", z)
print("Prediction:", prediction)
```

Calculate loss

```python
actual = 1

epsilon = 1e-15
prediction = np.clip(prediction, epsilon, 1 - epsilon)

loss = -(
    actual * np.log(prediction) +
    (1 - actual) * np.log(1 - prediction)
)

print("Actual:", actual)
print("Prediction:", prediction)
print("Loss:", loss)
```

Show the complete flow

```python
print("Input:", x)
print("Weights:", weights)
print("Bias:", bias)
print("Weighted Sum:", z)
print("Prediction:", prediction)
print("Actual:", actual)
print("Loss:", loss)
```
