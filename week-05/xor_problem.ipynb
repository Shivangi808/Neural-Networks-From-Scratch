```python
import numpy as np
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 1, 1, 0])
```

```python
import matplotlib.pyplot as plt

plt.scatter(X[y == 0, 0], X[y == 0, 1], label="Class 0")
plt.scatter(X[y == 1, 0], X[y == 1, 1], label="Class 1")

plt.xlabel("x₁")
plt.ylabel("x₂")
plt.title("XOR Problem")
plt.legend()
plt.grid()
plt.show()
```
