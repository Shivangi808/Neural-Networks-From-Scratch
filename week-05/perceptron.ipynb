```python
import numpy as np
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 0, 0, 1])
```

```python
weights = np.zeros(2)
bias = 0

print("Weights:", weights)
print("Bias:", bias)
```

```python
z = np.dot(X, weights) + bias

print(z)
```

```python
def step_function(z):
    return np.where(z >= 0, 1, 0)
```

```python
predictions = step_function(z)

print(predictions)
```
