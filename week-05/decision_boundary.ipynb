```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([
    [1, 1],
    [1, 2],
    [2, 1],
    [4, 4],
    [4, 5],
    [5, 4]
])

y = np.array([0, 0, 0, 1, 1, 1])
```

```python
plt.scatter(X[y == 0, 0], X[y == 0, 1], label="Class 0")
plt.scatter(X[y == 1, 0], X[y == 1, 1], label="Class 1")

plt.xlabel("x₁")
plt.ylabel("x₂")
plt.title("Linearly Separable Data")
plt.legend()
plt.grid()
plt.show()
```
