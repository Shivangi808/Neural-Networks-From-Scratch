```python
import numpy as np
```

```python
def step(z):
    return np.where(z >= 0, 1, 0)
```

Sigmoid

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

Tanh

```python
def tanh(z):
    return np.tanh(z)
```

ReLU

```python
def relu(z):
    return np.maximum(0, z)
```

Leaky ReLU

```python
def leaky_relu(z, alpha=0.01):
    return np.where(z > 0, z, alpha * z)
```

Softmax

```python
def softmax(z):
    exp_z = np.exp(z - np.max(z))
    return exp_z / np.sum(exp_z)
```

```python
z = np.array([-3, -2, -1, 0, 1, 2, 3])
```

```python
print("Step:", step(z))
print("Sigmoid:", sigmoid(z))
print("Tanh:", tanh(z))
print("ReLU:", relu(z))
print("Leaky ReLU:", leaky_relu(z))
```
