```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def step(z):
    return np.where(z >= 0, 1, 0)


def sigmoid(z):
    return 1 / (1 + np.exp(-z))


def tanh(z):
    return np.tanh(z)


def relu(z):
    return np.maximum(0, z)


def leaky_relu(z, alpha=0.01):
    return np.where(z > 0, z, alpha * z)
```

```python
x = np.linspace(-10, 10, 400)
```

Plot Step Function

```python
plt.plot(x, step(x))

plt.title("Step Function")
plt.xlabel("z")
plt.ylabel("f(z)")
plt.grid()

plt.show()
```

Plot Sigmoid

```python
plt.plot(x, sigmoid(x))

plt.title("Sigmoid Function")
plt.xlabel("z")
plt.ylabel("f(z)")
plt.grid()

plt.show()
```

Plot Tanh

```python
plt.plot(x, tanh(x))

plt.title("Tanh Function")
plt.xlabel("z")
plt.ylabel("f(z)")
plt.grid()

plt.show()
```

Plot ReLU

```python
plt.plot(x, relu(x))

plt.title("ReLU Function")
plt.xlabel("z")
plt.ylabel("f(z)")
plt.grid()

plt.show()
```

Plot Leaky ReLU

```python
plt.plot(x, leaky_relu(x))

plt.title("Leaky ReLU Function")
plt.xlabel("z")
plt.ylabel("f(z)")
plt.grid()

plt.show()
```

```python
plt.plot(x, step(x), label="Step")
plt.plot(x, sigmoid(x), label="Sigmoid")
plt.plot(x, tanh(x), label="Tanh")
plt.plot(x, relu(x), label="ReLU")
plt.plot(x, leaky_relu(x), label="Leaky ReLU")

plt.title("Activation Functions Comparison")
plt.xlabel("z")
plt.ylabel("f(z)")
plt.legend()
plt.grid()

plt.show()
```

## Observations

- Step function produces a discrete output of 0 or 1.
- Sigmoid produces smooth values between 0 and 1.
- Tanh produces values between -1 and 1.
- ReLU outputs 0 for negative inputs and the input itself for positive inputs.
- Leaky ReLU allows a small negative output for negative inputs.
- Activation functions introduce non-linearity into neural networks.
