Import NumPy

```python
import numpy as np
```

Define the functions

```python
x = 2
w = 3

z = w * x
loss = z ** 2

print("x:", x)
print("w:", w)
print("z:", z)
print("loss:", loss)
```

Calculate dL/dz

```python
dL_dz = 2 * z

print("dL/dz:", dL_dz)
```

Calculate dz/dw

```python
dz_dw = x

print("dz/dw:", dz_dw)
```

Apply the Chain Rule

```python
dL_dw = dL_dz * dz_dw

print("dL/dw:", dL_dw)
```

Complete calculation

```python
print("Forward Pass")
print("z =", z)
print("Loss =", loss)

print("\nBackward Pass")
print("dL/dz =", dL_dz)
print("dz/dw =", dz_dw)
print("dL/dw =", dL_dw)
```

Backpropagation applies the chain rule repeatedly to move gradients backward through a neural network.
