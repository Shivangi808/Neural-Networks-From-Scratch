Import NumPy

```python
import numpy as np
```

Define the data and parameters

```python
x = 2.0
w = 0.5
b = 1.0
y = 4.0

learning_rate = 0.1

print("Input:", x)
print("Weight:", w)
print("Bias:", b)
print("Actual:", y)
print("Learning Rate:", learning_rate)
```

Forward Propagation

```python
z = w * x + b
prediction = z

print("Weighted Sum:", z)
print("Prediction:", prediction)
```

Calculate Loss

```python
loss = 0.5 * (y - prediction) ** 2

print("Actual:", y)
print("Prediction:", prediction)
print("Loss:", loss)
```

Start Backpropagation

```python
dL_dprediction = prediction - y

print("dL/dprediction:", dL_dprediction)
```

Gradient of Weight

```python
dprediction_dw = x

dL_dw = dL_dprediction * dprediction_dw

print("dL/dw:", dL_dw)
```

Gradient of Bias

```python
dprediction_db = 1

dL_db = dL_dprediction * dprediction_db

print("dL/db:", dL_db)
```

Update Weight and Bias

```python
w = w - learning_rate * dL_dw
b = b - learning_rate * dL_db

print("Updated weight:", w)
print("Updated bias:", b)
```

Complete Training Loop

```python
x = 2.0
w = 0.5
b = 1.0
y = 4.0

learning_rate = 0.1

for step in range(20):

    # Forward propagation
    prediction = w * x + b

    # Calculate loss
    loss = 0.5 * (y - prediction) ** 2

    # Backpropagation
    dL_dprediction = prediction - y

    dL_dw = dL_dprediction * x
    dL_db = dL_dprediction

    # Gradient descent
    w = w - learning_rate * dL_dw
    b = b - learning_rate * dL_db

    print(
        f"Step {step + 1}: "
        f"Prediction = {prediction:.4f}, "
        f"Loss = {loss:.4f}"
    )
```

Store Loss

```python
x = 2.0
w = 0.5
b = 1.0
y = 4.0

learning_rate = 0.1
steps = 20

loss_history = []

for step in range(steps):

    # Forward propagation
    prediction = w * x + b

    # Loss
    loss = 0.5 * (y - prediction) ** 2
    loss_history.append(loss)

    # Backpropagation
    dL_dprediction = prediction - y
    dL_dw = dL_dprediction * x
    dL_db = dL_dprediction

    # Update parameters
    w = w - learning_rate * dL_dw
    b = b - learning_rate * dL_db

print("Initial loss:", loss_history[0])
print("Final loss:", loss_history[-1])
```

Visualize Loss

```python
import matplotlib.pyplot as plt

plt.plot(loss_history, marker="o")

plt.xlabel("Training Step")
plt.ylabel("Loss")
plt.title("Loss During Training")
plt.grid()

plt.show()
```

Final Parameters

```python
print("Final weight:", w)
print("Final bias:", b)
print("Final prediction:", w * x + b)
print("Final loss:", loss_history[-1])
```
