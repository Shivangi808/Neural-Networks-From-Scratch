```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def function(x):
    return x ** 2
```

```python
def derivative(x):
    return 2 * x
```

```python
def gradient_descent(start, learning_rate, steps):
    x = start
    history = [x]

    for _ in range(steps):
        gradient = derivative(x)
        x = x - learning_rate * gradient
        history.append(x)

    return history
```

```python
start = 5
learning_rate = 0.1
steps = 20

history = gradient_descent(
    start,
    learning_rate,
    steps
)

print("Final x:", history[-1])
print("Final function value:", function(history[-1]))
```

```python
for i, x in enumerate(history):
    print(f"Step {i}: x = {x:.6f}, f(x) = {function(x):.6f}")
```

```python
x_values = np.linspace(-6, 6, 400)
y_values = function(x_values)

plt.plot(x_values, y_values)

history = np.array(history)

plt.plot(
    history,
    function(history),
    marker="o"
)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Gradient Descent")
plt.grid()

plt.show()
```
