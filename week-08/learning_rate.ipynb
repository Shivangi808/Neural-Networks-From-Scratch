```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def function(x):
    return x ** 2


def derivative(x):
    return 2 * x
```

```python
def gradient_descent(start, learning_rate, steps):
    x = start
    history = [x]

    for _ in range(steps):
        gradient = derivative(x)
        x = x - learning_rate * gradient
        history.append(x)

    return history
```

```python
start = 5
steps = 20

learning_rates = [0.01, 0.1, 0.5, 1.1]

for lr in learning_rates:
    history = gradient_descent(start, lr, steps)

    print(
        f"Learning rate: {lr}, "
        f"Final x: {history[-1]:.6f}"
    )
```
